```python
travel_df = pd.DataFrame({
    'Name': ['Aman', 'Rahul', 'Priya', 'Neha', 'Karan'],
    'Destination': ['Delhi', 'Goa', 'Mumbai', 'Jaipur', 'Pune']
})


travel_df['Code'] = np.arange(5)

print("After np.arange():")
print(travel_df)


codes = pd.Series(
    [101, 104],
    index=[1, 4]
)


travel_df['Code'] = codes

print("\nAfter Series assignment:")
print(travel_df)
```

```python
school_df = pd.DataFrame({
    'Student': ['Aman', 'Rahul', 'Priya', 'Neha'],
    'Class': [10, 10, 11, 11],
    'Section': ['A', 'B', 'A', 'B']
})


new_school_df = school_df.reindex(
    columns=['Student', 'Class', 'Section', 'House']
)

print("Before assigning House:")
print(new_school_df)

new_school_df['House'] = 'Blue House'

print("\nAfter assigning House:")
print(new_school_df)
```

```python
inventory_df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Mouse', 'Keyboard'],
    'Quantity': [10, 20, 15, 30, 25],
    'Price': [60000, 30000, 25000, 800, 1200]
})


supplier = pd.Series(
    ['ABC Ltd', 'XYZ Ltd', 'PQR Ltd'],
    index=[0, 2, 4]
)


inventory_df['Supplier'] = supplier

print("After adding Supplier:")
print(inventory_df)


inventory_df = inventory_df.drop('Supplier', axis=1)

print("\nAfter deleting Supplier:")
print(inventory_df)
```
